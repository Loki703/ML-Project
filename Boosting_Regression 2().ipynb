```python
import pandas as pd
import numpy as np
from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.ensemble import GradientBoostingRegressor,AdaBoostRegressor
from sklearn.metrics import r2_score
from sklearn.linear_model import LinearRegression,Lasso,ElasticNet,Ridge
from sklearn.ensemble import RandomForestRegressor
```

```python
df=load_diabetes()
```

```python
df
```

```python
x=pd.DataFrame(data=df.data,columns=df.feature_names)
```

```python
x
```

```python
y=pd.DataFrame(df["target"])
```

```python
y
```

```python
y = np.ravel(y)
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.20,random_state=32)
```

## Ada Boosting --> Regression

```python
%%time
Ada_boost=AdaBoostRegressor(n_estimators=250)
Ada_boost.fit(x_train,y_train)
y_train_pred=Ada_boost.predict(x_train)
print("Train Score :",r2_score(y_train,y_train_pred))
y_test_pred=Ada_boost.predict(x_test)
print("Test Score :",r2_score(y_test,y_test_pred))
```

## Gradient Boosting --> Regression

```python
%%time
Grad_boost=GradientBoostingRegressor(n_estimators=150)
Grad_boost.fit(x_train,y_train)
y_train_pred=Grad_boost.predict(x_train)
print("Train Score :",r2_score(y_train,y_train_pred))
y_test_pred=Grad_boost.predict(x_test)
print("Test Score :",r2_score(y_test,y_test_pred))
```

## Linear Regression

```python
%%time
Linearregression=LinearRegression()
Linearregression.fit(x_train,y_train)
y_train_pred=Linearregression.predict(x_train)
y_test_pred=Linearregression.predict(x_test)
print("Train Score :",r2_score(y_train,y_train_pred))
print("Test Score :",r2_score(y_test,y_test_pred))
```

## Random Forest ---> Regression

```python
%%time
RF_model= RandomForestRegressor(n_estimators=150)
RF_model.fit(x_train,y_train)
y_train_pred=RF_model.predict(x_train)
y_test_pred=RF_model.predict(x_test)
print("Train Score :",r2_score(y_train,y_train_pred))
print("Test Score :",r2_score(y_test,y_test_pred))
```

## To Avoid Overfitting in Linear Regression ---> We can go with regularization Algorithms

```python
from sklearn.linear_model import LinearRegression,Lasso,ElasticNet,Ridge
```

## Using Lasso regression

```python
Lasso_regression=Lasso()
Lasso_regression.fit(x_train,y_train)
y_train_pred=Lasso_regression.predict(x_train)
print("r2_score of train data :",r2_score(y_train,y_train_pred))
y_test_pred=Lasso_regression.predict(x_test)
print("r2_score of test data :",r2_score(y_test,y_test_pred))
```

## Using Elastic Net

```python
Elastic_regression=ElasticNet()
Elastic_regression.fit(x_train,y_train)
y_train_pred=Elastic_regression.predict(x_train)
print("r2_score of train data :",r2_score(y_train,y_train_pred))
y_test_pred=Elastic_regression.predict(x_test)
print("r2_score of test data :",r2_score(y_test,y_test_pred))
```

## Using Ridge Regression

```python
Ridge_regression=Ridge()
Ridge_regression.fit(x_train,y_train)
y_train_pred=Ridge_regression.predict(x_train)
print("r2_score of train data :",r2_score(y_train,y_train_pred))
y_test_pred=Ridge_regression.predict(x_test)
print("r2_score of test data :",r2_score(y_test,y_test_pred))
```
